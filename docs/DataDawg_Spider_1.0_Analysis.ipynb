```python
import pandas as pd

# Import the xlsx file as a pandas DataFrame
df = pd.read_excel('./spider_benchmark_results.xlsx')
display(df)
```

```python
import matplotlib.pyplot as plt

# Correct column names for df
df.columns = [col.lower().replace(' ', '_') for col in df.columns]

# Use record number for X axis values
x_values = list(range(1, len(df) + 1))

# Use correct accuracy columns for Y axis values
y_baseline = df['baseline_accuracy']
y_enhanced = df['enhanced_accuracy']

plt.figure(figsize=(12, 5))  # Increased width by 50%
plt.plot(x_values, y_baseline, marker='o', label='Baseline Accuracy')
plt.plot(x_values, y_enhanced, marker='s', label='Enhanced Accuracy')
plt.xlabel('Record Number')
plt.ylabel('Accuracy')
plt.title('Accuracy for Baseline and Enhanced by Record Number')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
